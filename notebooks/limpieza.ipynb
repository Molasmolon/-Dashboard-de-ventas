## 4. Limpieza.
## Conversión de fechas.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

df.dtypes
```

## 5. Creación de columnas nuevas.
## Extraer mes, año y calcular cuántos días tardó en enviarse cada orden..
 

```python
df['Mes'] = df['Order Date'].dt.month
df['Año'] = df['Order Date'].dt.year
df['Dias de Despacho'] = (df['Ship Date'] - df['Order Date']).dt.days

df[['Order Date', 'Ship Date', 'Mes', 'Año', 'Dias de Despacho']].head(10)
```

## 6. Agrupación de datos.
## Ventas por mes y año.

```python
ventas_por_mes = df.groupby(['Año', 'Mes'])['Sales'].sum().reset_index()
ventas_por_mes.columns = ['Año', 'Mes', 'Total Ventas']
ventas_por_mes.head(40)
```
