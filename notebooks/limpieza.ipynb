# Análisis de Ventas — Superstore.
## 1. Importación de librerías.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Carga del dataset

```python
df = pd.read_csv('../data/train.csv', encoding='latin1')
df.head()
```

## 3. Exploración inicial.
## Saber cuántas filas y columnas tiene el dataset.
## Ver los tipos de datos.
## Buscar valores nulos.

```python
df.shape
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

## 4. Limpieza.
## Conversión de fechas.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

df.dtypes
```

## 5. Creación de columnas nuevas.
## Extraer mes, año y calcular cuántos días tardó en enviarse cada orden..
 

```python
df['Mes'] = df['Order Date'].dt.month
df['Año'] = df['Order Date'].dt.year
df['Dias de Despacho'] = (df['Ship Date'] - df['Order Date']).dt.days

df[['Order Date', 'Ship Date', 'Mes', 'Año', 'Dias de Despacho']].head(10)
```

## 6. Agrupación de datos.
## Ventas por mes y año.

```python
ventas_por_mes = df.groupby(['Año', 'Mes'])['Sales'].sum().reset_index()
ventas_por_mes.columns = ['Año', 'Mes', 'Total Ventas']
ventas_por_mes.head(40)
```

## 7. Visualización rápida.
## Grafica de Ventas por mes.

```python
ventas_2017 = ventas_por_mes[ventas_por_mes['Año'] == 2017]

plt.figure(figsize=(12, 5))
plt.plot(ventas_2017['Mes'], ventas_2017['Total Ventas'], marker='o', color='steelblue')
plt.title('Ventas mensuales 2017')
plt.xlabel('Mes')
plt.ylabel('Total Ventas')
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```
